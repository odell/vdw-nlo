# ${}^{4}{\rm He}$

The interaction is defined in `helium4.py`. It is nonlocal, and so is the counterterm. We will use it to tune the scattering length, $a_0$, **and** the effective range, $r_0$, over a range of short-distance cutoffs, $R$.

```python
import numpy as np
from scipy import optimize
import matplotlib.pyplot as plt
import scienceplots

import helium4plusplus as he4
import lm2m2

plt.style.use('science')
gr = 4/3
h = 3
```

```python
# angular momentum
ELL = 0

# load lo coupling constants
Rs, glos, gnlos = np.loadtxt('datfiles/rg_flows/nonlocal_nlo_p.txt', unpack=True)
# number of short distance cutoffs
nR = Rs.size
```

```python
index = 0
R = Rs[index]
glo = glos[index]
gnlo = gnlos[index]

s = he4.NonlocalHelium4System2(R, ELL, 2, 6, 6)
```

```python
1000 * s.bound_state_spectrum(glo, gnlo)
```

```python
with open(f'datfiles/matrix_elements/nlo_pert_R_{R:.8e}_glo_{glo:.8e}_gnlo_{gnlo:.8e}.txt', 'w+') as f:
    f.write('# weight[i] (1/Å)  momenta[i] (1/Å)  momenta[j] (1/Å)  v_0[i, j] (K•Å^3)  v_1[i, j] (K•Å^3)\n')
    v0 = s.v_tilde + s.interaction.counterterm.gen(glo, 0)
    v1 = s.interaction.counterterm.gen(0, gnlo)
    for i in range(s.q.size):
        for j in range(s.q.size):
            f.write(f'{s.wq[i]:.8e}  {s.q[i]:.8e}  {s.q[j]:.8e}  {v0[i, j]:.8e}  {v1[i, j]:.8e}\n')
```
