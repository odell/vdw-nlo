```python
import numpy as np
import matplotlib.pyplot as plt
import helium4plusplus as he4
import scienceplots

plt.style.use('science')
```

```python
nonpert = np.loadtxt('datfiles/rg_flows/nonlocal_nlo_np.txt')
pert = np.loadtxt('datfiles/rg_flows/nonlocal_nlo_p.txt')
```

```python
fig, ax = plt.subplots(dpi=300)
fig.patch.set_facecolor('white')

ax.plot(nonpert[:, 0]/he4.BETA6, nonpert[:, 1], marker='o', linestyle='', label='LO')
ax.plot(nonpert[:, 0]/he4.BETA6, nonpert[:, 2], marker='s', linestyle='', label='NLO')

ax.legend()
ax.set_xlabel(r'$R/\beta_6$')
ax.set_ylabel(r'Coupling Strength');

plt.savefig('figures/nonpert_nlo_nonlocal_rg_flow.png', dpi=300)
# ax.set_yscale('log');
```

```python
fig, ax = plt.subplots(dpi=300)
fig.patch.set_facecolor('white')

ax.plot(pert[:, 0]/he4.BETA6, pert[:, 1], marker='o', linestyle='', label='LO')
ax.plot(pert[:, 0]/he4.BETA6, pert[:, 2], marker='s', linestyle='', label='NLO')

ax.legend()
ax.set_xlabel(r'$R/\beta_6$')
ax.set_ylabel(r'Coupling Strength');

ax.set_ylim([-500, 500])

plt.savefig('figures/pert_nlo_nonlocal_rg_flow.pdf', dpi=300)
# ax.set_yscale('log');
```

```python
def sort_by_cutoff(d):
    ii = np.argsort(d[:, 0])
    return d[ii, :]
```

```python
# Arnas plots
b3_lo = sort_by_cutoff(np.loadtxt('datfiles/deltuva_nonlocal_lo_bgs+.txt'))
b3_nlo = sort_by_cutoff(np.loadtxt('datfiles/deltuva_nonlocal_nlo_bgs+.txt'))

b3_nlo_new = sort_by_cutoff(np.loadtxt('datfiles/x_full'))
b3_nlo_new_short = sort_by_cutoff(np.loadtxt('datfiles/x_short'))
```

```python
fig, ax = plt.subplots(dpi=300)
fig.patch.set_facecolor('white')

ax.plot(b3_lo[:, 0], b3_lo[:, 1], linestyle='-', label=r'$R/10$ LO')
ax.plot(b3_nlo[:, 0], b3_nlo[:, 1], linestyle='--', label=r'$R/10$ NLO')

ax.plot(b3_nlo_new[:, 0], b3_nlo_new[:, 1], linestyle='-.', label=r'$R$ NLO')
ax.plot(b3_nlo_new_short[:, 0], b3_nlo_new_short[:, 1], linestyle=':', label=r'$R$ NLO')

ax.legend()
ax.set_xlabel(r'$R$')
ax.set_ylabel(r'$B_3$');
```
